```python
user_query = "요즘 주택 청약 관련 소식이 궁금해요" 
contents_keyword = '주택 청약'  

# 기사 검색 및 원문 기반 재정렬
articles = search_naver_news(contents_keyword)  
article_links = [link for _, link in articles]
 
# top_k 개의 user_query와 관련도가 높은순으로 정렬된 기사 원문 리스트 
ranked_articles = rerank_articles_by_scores(user_query, article_links, top_k=5)  # 정렬할 상위 문서 개수 k 지정
```

```python
def doc_rank(func):
    def wrapper(n):
        print(f'Rerank된 문서 중 {n+1}위 기사의 요약입니다.')
        # 재정렬된 기사 원문 ranked_article에서 n번째 기사를 요약 
        result = func(n)   
        return result
    return wrapper


@doc_rank
def article_summary(n: int) -> str:
    """입력받은 기사 원문을 전처리하고 주요 내용을 요약합니다.""" 
    ranked_articles[n][1]
    response = client.chat.completions.create(
        model="gpt-4o",  
        messages=[
            {"role": "system", "content": f"입력받은 기사 원문 {ranked_articles}에서 특수 문자를 제거하고 주요 내용을 요약하세요."},
        ],
        max_tokens=800,
        temperature=0
    )

    cleaned_article = response.choices[0].message.content 
    
    return cleaned_article


article_summary(0)  
```
